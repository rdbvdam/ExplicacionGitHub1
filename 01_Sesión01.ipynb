### 1. Tipos dinámicos
Python es de [tipado dinámico](https://es.wikipedia.org/wiki/Tipado_din%C3%A1mico), lo que significa que una variable puede almacenar datos de cualquier tipo, y ese tipo puede cambiar .

Por ejemplo, si creo una variable numérica, posteriormente puedo cambiar esa variable a una cadena de texto:

```python
suscriptores_activos = 45200
suscriptores_activos = "Cuarenta y cinco mil"
```

Que algo se pueda hacer no tiene por qué ser, necesariamente, una buena idea. En casi todas las circunstancias es una __mala idea__ cambiar el tipo de una variable. Lo _correcto_ suele ser, simplemente, crear una nueva variable. Por ejemplo:

```python
suscriptores_activos = 45200
suscriptores_descripcion = "Cuarenta y cinco mil"
```

Los lenguajes que no son de tipado dinámico son de tipado estático, como Go y TypeScript. En un lenguaje de tipado estático, si intentas asignar un valor a una variable de tipo incorrecto, obtendrás un error de compilación y el programa no se ejecutará.

Si Python fuera de tipado estático, el primer ejemplo anterior no permitiría la segunda línea, `suscriptores_activos = "Cuarenta y cinco mil"`. El ordenador daría un error del tipo _you can't assign a string value ("Cuarenta y cinco mil") to a number variable (suscriptores_activos)_.

### 2. Sugerencias de tipo o _Type hints_
Cuando un programa es pequeño, normalmente se pueden recordar los tipos de las variables. Pero a medida que los programas crecen es conveniente indicar el tipo de una variable para facilitar su lectura y aprovechar las funciones de los editores de código, que pueden indicar errores en función de estas sugerencias de tipo:

¿Es `alias` un str?
¿Qué tipo de valor numérico admite la media de la audiencia?
Las sugerencias de tipo nos permiten escribir esas expectativas directamente en nuestro código:


```python
# None
alias: str | None = "None"

# Float (es un int)
espectadores_media: float = 78500

# Entero (es un str)
suscriptores_activos: int = "45200"

# Booleano (es un str)
emite_hoy: bool = 'True'
```

Las sugerencias de tipo no hacen que Python deje de ser Python . Sigue siendo un lenguaje de tipado dinámico y, como se puede ver en el ejemplo anterior, no rechazará automáticamente un valor incorrecto solo porque una sugerencia de tipo lo indique, pero probablemente vuestro editor os marcará el error para corregirlo.

Las sugerencias de tipo sirven para:

* Hacer que el código sea más fácil de leer.
* Ayudar a tu editor a autocompletar y avisar de los errores.
* Facilitar la detección de errores antes de ejecutar el código.

### 3. Listas y tuplas: Manipulación de secuencias

Las __listas__ son colecciones ordenadas y mutables de elementos. Pueden contener cualquier tipo de datos. Constituyen la estructura nativa fundamental para almacenar series de observaciones antes de trabajar con librerías especializadas como `numpy` o `pandas`.
A continuación, analizaremos una muestra de canciones con mayor número de reproducciones diarias en Spotify España.

```python
### 5. Trabajo con listas: mutabilidad, indexación y troceado (slicing)

"""
A veces, al crear listas manualmente, puede resultar difícil
leerlas si todos los elementos están en la misma línea de código.
Podemos declarar la lista usando varias líneas:
"""
canciones: list[str] = [
    "BbY WOW",
    "La Graciosa",
    "De Lejitos - Remix",
    "Al Golpito"
]

# O en una sola línea:
reproducciones_millones: list[float] = [0.535, 0.444, 0.361, 0.311]

# Acceso por índice e indexación negativa (último elemento)
print(f"Canción número 1 en España: {canciones[0]}")
print(f"Última canción de la lista: {canciones[-1]}")

# Tamaño de la lista
print(f"Lista con las {len(canciones)} canciones con más reproducciones:")
for cancion in canciones:
  print(cancion)

# Slicing: Obtener las dos primeras canciones
top_2: list[str] = canciones[:2]
print(f"Top 2 de la lista: {top_2}")

# Actualizar contenido de las listas
reproducciones_millones[0] = 0.602

# Añadir elementos a la lista
canciones.append("Vamo a vel")
reproducciones_millones.append(0.305)

# Eliminar (extraer) elementos de las listas
cancion_eliminada: str = canciones.pop()
# se le puede pasar el índice a eliminar que, en este ejemplo, se pierde:
reproducciones_millones.pop(-1)
print(f"Canción eliminada: {cancion_eliminada}")
print(f"Valor de reproducciones último: {reproducciones_millones[-1]}")

# Operaciones de agregación nativas
total_reproducciones: float = sum(reproducciones_millones)
media_reproducciones: float = total_reproducciones / len(reproducciones_millones)
# Mostramos los valores con dos decimales:
print(f"Total reproducciones acumuladas: {total_reproducciones:.2f} millones")
print(f"Promedio de reproducciones por tema: {media_reproducciones:.2f} millones")
```

Las __tuplas__ son estructuras de datos
ordenadas pero inmutables —una vez creadas, sus elementos no pueden modificarse, añadirse ni eliminarse—. En _machine learning_ y desarrollo de software son fundamentales para almacenar registros fijos (como coordenadas o configuraciones) y para el retorno de múltiples valores en funciones.
Puedes pensar en una tupla como una lista de tamaño fijo. Las tuplas se crean con paréntesis:


```python
mi_tupla: tuple[str, int, bool] = ("Esto es una tupla", 45, True)
print(mi_tupla[0])
print(mi_tupla[1])
print(mi_tupla[2])
```

Si bien generalmente se considera una mala práctica almacenar elementos de diferentes tipos en una lista, esto no representa un problema con las tuplas. Debido a que tienen un tamaño fijo, es fácil llevar un registro de qué índices almacenan qué tipos de datos.
Las tuplas se utilizan a menudo para almacenar grupos muy pequeños de datos (como 2 o 3 elementos). Por ejemplo, se podría usar una tupla para almacenar el nombre y el nivel alcanzado por un jugador: `player = ("Fideas", 5)`.

Existe un caso especial para crear tuplas de un solo elemento. Hay que incluir una coma al final para que Python sepa que es una tupla y no paréntesis normales: `player = ("Crisss",)`.

Dado que las tuplas almacenan sus datos, se pueden guardar varias tuplas en una lista. Al igual que con otros datos, cada tupla se separa por una coma. Para acceder a una lista de tuplas, el primer índice selecciona la tupla a la que se desea acceder, y el segundo, un valor dentro de ella.


```python
# Creación de una lista de tuplas con type hints
# Cada elemento es una tupla: (nombre: str, nivel: int)
jugadores: list[tuple[str, int]] = [
    ("Faker", 5),
    ("Caps", 4),
    ("Rookie", 4),
    ("Elyoya", 3),
    ("Jojopyun", 2),
]

# 1. Acceso por índice
print(f"Primer jugador: {jugadores[0][0]} (Nivel {jugadores[0][1]})")

# 2. Acceso al último elemento con índice negativo mediante desempaquetado
ultimo_nombre, ultimo_nivel = jugadores[-1]
print(f"Último jugador: {ultimo_nombre} (Nivel {ultimo_nivel})")

# 3. Recorrer la lista desempaquetando las tuplas directamente
print("\n--- Lista completa de jugadores ---")
for nombre, nivel in jugadores:
    print(f"- {nombre}: Nivel {nivel}/5")

```

Veamos un ejemplo de las operaciones típicas con tuplas:

```python
### 5. Trabajo con tuplas: inmutabilidad, indexación y desempaquetado

# Definición de una tupla:
# (Nombre del evento, Ciudad, Aforo máximo, Precio entrada €)
evento_velada: tuple[str, str, int, float] = (
    "La Velada del Año IV",
    "Madrid",
    80000,
    55.0
)

# 1. Acceso por índice e indexación negativa
print(f"Evento: {evento_velada[0]}")
print(f"Precio de la entrada básica: {evento_velada[-1]} €")

# 2. Troceado
localizacion = evento_velada[:2]
print(f"Datos de localización (Nombre y Ciudad): {localizacion}")

# 3. Demostración de inmutabilidad
# Intentar modificar un elemento de la tupla provocará un error de tipo (TypeError)
try:
  evento_velada[2] = 85000
except TypeError as error:
  print(
    f"\n[Demostración de inmutabilidad] No se puede modificar la tupla: {error}"
    )

# 4. Desempaquetado de tuplas
nombre, ciudad, aforo, precio = evento_velada
print(f"\n--- Ficha desempaquetada ---")
print(
    f"El evento «{nombre}» se celebrará en {ciudad}\ncon un aforo de {aforo:,} personas."
    )

# 5. Retorno múltiple de valores en funciones usando tuplas
def calcular_metricas_estadio(aforo_total: int, entradas_vendidas: int) -> tuple[float, float]:
  """Calcula el porcentaje de ocupación y la recaudación estimada."""
  porcentaje_ocupacion: float = (entradas_vendidas / aforo_total) * 100
  recaudacion: float = entradas_vendidas * 55.0
  # Devuelve implícitamente una tupla: (porcentaje, recaudación)
  return porcentaje_ocupacion, recaudacion
porcentaje, ingresos = calcular_metricas_estadio(evento_velada[2], 75000)
print(f"\n--- Estadísticas del estadio ---")
print(f"Porcentaje de ocupación: {porcentaje:.2f}%")
print(f"Recaudación estimada: {ingresos:.2f} €")
```

### 4. Diccionarios: Mapeos clave-valor

Mientras que las listas y tuplas organizan la información mediante índices numéricos secuenciales (0, 1, 2...), los **diccionarios (`dict`)** asocian datos a través de claves únicas.

Son, por tanto, colecciones mutables formadas por pares `clave: valor`. Puedes pensar en ellos como una agenda telefónica: no buscas el "teléfono número 5", sino que buscas por el "nombre del contacto" (la clave) para obtener su "número" (el valor).

En análisis de datos, los diccionarios son fundamentales porque son el paso previo natural a los archivos JSON o a las filas de un _DataFrame_ en `pandas`. Además, el acceso a sus elementos es instantáneo, sin importar lo grandes que sean.

```python
### 6. Trabajo con diccionarios: accesos, modificaciones e iteración

# Definimos el estado de una estación del sistema BIKI de Valladolid.
# Las claves (strings) representan el ID del anclaje, y los valores el tipo de bici.
estacion_plaza_mayor: dict[str, str | None] = {
    "P01": "eléctrica",
    "P02": "mecánica",
    "P03": None,  # Anclaje libre
    "P04": "eléctrica"
}

# 1. Acceso a elementos por clave
print(f"En el anclaje P01 hay una bicicleta {estacion_plaza_mayor['P01']}")

# 2. Acceso seguro con .get()
# Si intentamos acceder a una clave que no existe con corchetes,
# Python dará un KeyError.
# .get() nos permite devolver un valor por defecto si la clave no existe.
estado_p05 = estacion_plaza_mayor.get("P05", "Anclaje no instalado")
print(f"Estado del anclaje P05: {estado_p05}")

# 3. Modificación e inserción
# Un usuario devuelve una bici mecánica
estacion_plaza_mayor["P03"] = "mecánica"

# Se amplía la estación instalando un nuevo anclaje
estacion_plaza_mayor["P05"] = "eléctrica"

# 4. Métodos para iterar: .keys(), .values(), .items()
print("\n--- Inventario actualizado de la estación ---")
for anclaje, tipo in estacion_plaza_mayor.items():
    if tipo is not None:
        print(f"-> {anclaje}: BIKI {tipo}")
```

### 5. Conjuntos (Sets): Unicidad y operaciones de teoría de conjuntos

Los **conjuntos (`set`)** son colecciones desordenadas de elementos **únicos**. A diferencia de las listas, un `set` no puede contener elementos duplicados y no mantiene un orden específico (por lo que no puedes acceder a ellos con índices numéricos como `set[0]`).

Se declaran usando llaves `{}` al igual que los diccionarios, pero sin pares clave-valor (solo elementos sueltos).

Son excepcionalmente útiles para dos tareas en el tratamiento de datos:
1. **Eliminar duplicados** de forma rápida al transformar una lista en un set.
2. **Realizar operaciones algebraicas** (unión, intersección, diferencia), igual que en los diagramas de Venn matemáticos.

```python
### 7. Trabajo con sets: filtrado de duplicados y álgebra de conjuntos

# 1. Unicidad (Casting de lista a set)
loot_cofre_duplicados: list[str] = ["Poción", "Oro", "Espada", "Poción", "Poción", "Oro"]
loot_unico: set[str] = set(loot_cofre_duplicados)
print(f"De {len(loot_cofre_duplicados)} objetos, los únicos son: {loot_unico}")

# 2. Operaciones de conjuntos
jugadores_clan_alfa = {"Thorin", "Legolas", "Gimli", "Aragorn"}
jugadores_conectados = {"Legolas", "Frodo", "Aragorn", "Sam"}

# INTERSECCIÓN (&): Elementos en ambos conjuntos
# (¿Quién del clan alfa está conectado?)
alfa_conectados = jugadores_clan_alfa & jugadores_conectados
print(f"\nMiembros del Clan Alfa online: {alfa_conectados}")

# DIFERENCIA (-): Elementos en el primero que NO están en el segundo
# (¿Quién falta por conectarse?)
alfa_desconectados = jugadores_clan_alfa - jugadores_conectados
print(f"Miembros del Clan Alfa offline: {alfa_desconectados}")

# UNIÓN (|): Todos los elementos de ambos conjuntos sin repetir
# (Total de jugadores implicados)
todos_los_jugadores = jugadores_clan_alfa | jugadores_conectados
print(f"Total de jugadores interactuando: {todos_los_jugadores}")
```

## Bloque 3 Clases y Objetos: Programación Orientada a Objetos (POO)

Hasta ahora hemos usado tipos de datos predefinidos por Python (listas, diccionarios, enteros). Sin embargo, cuando nuestros programas crecen, necesitamos representar entidades del mundo real de forma más natural y estructurada.

La **Programación Orientada a Objetos (POO)** nos permite crear nuestros propios tipos de datos usando **Clases**.

* **Clase:** Es la plantilla abstracta. Define qué datos tendrá el objeto y qué acciones puede realizar.
* **Objeto (la instancia):** Es una representación concreta creada a partir de esa clase.

Cada objeto encapsula un estado interno (**atributos** o variables) y unos comportamientos (**métodos** o funciones). El método especial `__init__` se ejecuta automáticamente al crear un nuevo objeto, y la palabra clave `self` hace referencia al objeto mismo que está ejecutando la acción.

```python
### 8. Trabajo con clases y objetos: Atributos, métodos y estado interno

# Definimos la clase (el molde) para las tarjetas de transporte de AUVASA
class TarjetaAUVASA:
    # El método __init__ inicializa los atributos del objeto al crearlo
    def __init__(self, titular: str, tipo_abono: str, saldo_inicial: float):
        self.titular = titular
        self.tipo_abono = tipo_abono
        self.saldo = saldo_inicial
        self.viajes_realizados = 0  # Atributo que empieza en 0 por defecto

    # Método para realizar una acción (comportamiento)
    def registrar_viaje(self):
        # Definimos el precio según el perfil del estudiante
        if self.tipo_abono == "Bono Joven":
            precio: float = 0.55
        else:
            precio = 0.75

        if self.saldo >= precio:
            self.saldo -= precio
            self.viajes_realizados += 1
            print(f"Viaje validado. Buen viaje, {self.titular}. Saldo restante: {self.saldo:.2f} €")
        else:
            print(f"❌ Saldo insuficiente ({self.saldo:.2f} €). ¡Recarga tu tarjeta!")

    # Método para modificar el estado (recargar saldo)
    def recargar(self, cantidad: float):
        self.saldo += cantidad
        print(f"✅ Recarga de {cantidad:.2f} € completada. Nuevo saldo: {self.saldo:.2f} €")


# 1. Instanciación: Creamos objetos reales a partir de la clase
tarjeta_pablo = TarjetaAUVASA("Pablo", "Bono Joven", 1.00)
tarjeta_profe = TarjetaAUVASA("Ángel", "Bono Ordinario", 5.00)

# 2. Uso de métodos (Interactuando con los objetos)
print("--- Movimientos en la Línea C1 ---")
tarjeta_pablo.registrar_viaje()
tarjeta_pablo.registrar_viaje()  # Esto debería dar error de saldo

print("\n--- Operaciones de mantenimiento ---")
tarjeta_pablo.recargar(5.00)
tarjeta_pablo.registrar_viaje()

# 3. Consulta directa de atributos
print(f"\nTotal de viajes realizados por {tarjeta_pablo.titular}: {tarjeta_pablo.viajes_realizados}")
```
